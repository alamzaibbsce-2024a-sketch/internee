```python
from transformers import pipeline

classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

print("Chatbot NLP model loaded successfully!")
```

```python
faq_data = {
    "What are the internship working hours?":
        "The internship working hours depend on the assigned internship schedule.",

    "How do I submit my project?":
        "You can submit your project through the internship task portal by following the submission instructions.",

    "Where can I find my internship tasks?":
        "Your assigned tasks can be found in the internship task portal.",

    "How do I contact my mentor?":
        "You can contact your assigned mentor through the communication method provided by the internship program.",

    "When will my project be reviewed?":
        "Projects are reviewed after submission according to the internship program's review process.",

    "How can I get my certificate?":
        "Complete the required internship projects and follow the program's certificate requirements.",

    "What should I do if I have a problem with my project?":
        "Review the project guidelines first, then contact your mentor or internship support if you still need help."
}

print("FAQ knowledge base created!")
print("Number of FAQs:", len(faq_data))
```

```python
def chatbot(user_question):
    questions = list(faq_data.keys())

    result = classifier(
        user_question,
        questions
    )

    best_question = result["labels"][0]
    confidence = result["scores"][0]

    if confidence >= 0.40:
        return faq_data[best_question]
    else:
        return "Sorry, I don't have enough information to answer that question."

print("Chatbot function created successfully!")
```

```python
question = "How do I submit my project?"

response = chatbot(question)

print("User:", question)
print("Chatbot:", response)
```
