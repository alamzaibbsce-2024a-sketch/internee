```python
from transformers import pipeline

print("Libraries imported successfully!")
```

```python
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)

print("Text-generation model loaded successfully!")
```

```python
prompt = """
Generate 5 technical interview questions for a beginner
Data Science intern. Questions should cover Python,
Pandas, machine learning, and data analysis.
"""

questions = generator(
    prompt,
    max_length=200,
    num_return_sequences=1
)

print(questions[0]["generated_text"])
```

```python
prompt = """
Generate 5 behavioral interview questions for a beginner
Data Science intern. Focus on teamwork, communication,
problem-solving, learning, and handling challenges.
"""

behavioral_questions = generator(
    prompt,
    max_length=200,
    num_return_sequences=1
)

print(behavioral_questions[0]["generated_text"])
```

```python
intern_profile = """
Beginner Data Science student.
Skills: Python, Pandas, NumPy, Matplotlib, and basic machine learning.
Experience: Completed beginner machine learning projects.
"""

job_role = "Data Science Intern"

prompt = f"""
Generate 5 customized interview questions for a candidate applying
for the role of {job_role}.

Candidate profile:
{intern_profile}

The questions should match the candidate's skills and the job role.
Include both technical and behavioral questions.
"""

custom_questions = generator(
    prompt,
    max_length=300,
    num_return_sequences=1
)

print(custom_questions[0]["generated_text"])
```

```python
job_description = """
We are looking for a Data Science Intern.
The candidate should have basic knowledge of Python,
Pandas, NumPy, data visualization, and machine learning.
The intern will help clean datasets, analyze data,
build basic machine learning models, and create reports.
Good communication and problem-solving skills are also required.
"""

prompt = f"""
Generate 5 interview questions for the following job.

Job Role:
{job_role}

Candidate Profile:
{intern_profile}

Job Description:
{job_description}

Create questions that match the candidate's skills
and the requirements of the job. Include both technical
and behavioral questions.
"""

job_questions = generator(
    prompt,
    max_length=400,
    num_return_sequences=1
)

print(job_questions[0]["generated_text"])
```

```python
technical_prompt = f"""
Generate 5 technical interview questions for the role of {job_role}.

Candidate Profile:
{intern_profile}

Job Description:
{job_description}

Focus on the technical skills and responsibilities mentioned
in the job description.
"""

behavioral_prompt = f"""
Generate 5 behavioral interview questions for the role of {job_role}.

Candidate Profile:
{intern_profile}

Job Description:
{job_description}

Focus on teamwork, communication, problem-solving,
learning ability, and handling challenges.
"""

technical_questions = generator(
    technical_prompt,
    max_length=300,
    num_return_sequences=1
)

behavioral_questions = generator(
    behavioral_prompt,
    max_length=300,
    num_return_sequences=1
)

print("===== TECHNICAL QUESTIONS =====")
print(technical_questions[0]["generated_text"])

print("\n===== BEHAVIORAL QUESTIONS =====")
print(behavioral_questions[0]["generated_text"])
```

```python
def generate_interview_questions(intern_profile, job_role, job_description):

    technical_prompt = f"""
    Generate 5 technical interview questions for the role of {job_role}.

    Candidate Profile:
    {intern_profile}

    Job Description:
    {job_description}

    Focus on the technical skills and responsibilities
    mentioned in the job description.
    """

    behavioral_prompt = f"""
    Generate 5 behavioral interview questions for the role of {job_role}.

    Candidate Profile:
    {intern_profile}

    Job Description:
    {job_description}

    Focus on teamwork, communication, problem-solving,
    learning ability, and handling challenges.
    """

    technical_questions = generator(
        technical_prompt,
        max_length=300,
        num_return_sequences=1
    )

    behavioral_questions = generator(
        behavioral_prompt,
        max_length=300,
        num_return_sequences=1
    )

    print("===== TECHNICAL QUESTIONS =====")
    print(technical_questions[0]["generated_text"])

    print("\n===== BEHAVIORAL QUESTIONS =====")
    print(behavioral_questions[0]["generated_text"])
```

```python
intern_profile = """
Beginner Data Science student with knowledge of Python,
Pandas, NumPy, Matplotlib, and basic machine learning.
"""

job_role = "Data Science Intern"

job_description = """
The intern will clean datasets, perform data analysis,
create visualizations, build basic machine learning models,
and communicate findings to the team.
"""

generate_interview_questions(
    intern_profile,
    job_role,
    job_description
)
```
